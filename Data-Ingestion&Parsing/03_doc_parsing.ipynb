```python
import os


os.makedirs("data/wordfiles", exist_ok = True)

with open("project_report.docx","rb") as f:
    content = f.read()

with open("data/wordfiles/report.docx", "wb") as f:
    f.write(content)
```

```python
from langchain_community.document_loaders import Docx2txtLoader, UnstructuredWordDocumentLoader
```

Method1: Using Docx2txtLoader:

```python
try:

    doc_loader = Docx2txtLoader("data/wordfiles/report.docx")

    docs = doc_loader.load()

    print(f"Total documents: {len(docs)}")
    print(f"The content preview of doc1 is: \n{docs[0].page_content[:300]}...")
    print(f"Meatadata of doc1: {docs[0].metadata}")

except Exception as e:
    print(f"Error: {e}")
```

Method2: Using UnstructuredWordDocumentLoader:

```python
try:

    unstruc_docx_loader = UnstructuredWordDocumentLoader("data/wordfiles/report2.docx", mode = "elements")

    unstruc_docs = unstruc_docx_loader.load()

    print(f"Total docs: {len(unstruc_docs)}")

    for i , doc in enumerate(unstruc_docs[:3]):
        print(f"Element: {i + 1}")
        print(f"type: {doc.metadata.get('category','unknown')}")
        print(f"Preview: {doc.page_content[:200]}...")

except Exception as e:
    print(f"Error: {e}")
```
